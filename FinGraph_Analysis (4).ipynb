```python
import pandas as pd
smurfing_data = []

for i in range(1, 21):
    smurfing_data.append({
        "transaction_id": f"SM{i:03d}",
        "sender_account": f"SMURF{i:03d}",
        "receiver_account": "TARGET01",
        "amount": 9900,
        "timestamp": f"2026-08-11 12:{i:02d}:00",
        "sender_ip": f"192.168.10.{i}"
    })

smurfing_df = pd.DataFrame(smurfing_data)

print(smurfing_df)
```

```python
id_counts = smurfing_df.groupby("receiver_account")["sender_account"].nunique()

print(id_counts)
```

```python
total_smurfing_amount = smurfing_df["amount"].sum()
average_smurfing_amount = smurfing_df["amount"].mean()

print("Total amount:", total_smurfing_amount)
print("Average transaction:", average_smurfing_amount)
```

```python
smurfing_risk_score = 0

# Many different senders
if id_counts["TARGET01"] >= 10:
    smurfing_risk_score += 40

# Transactions close to $10,000 threshold
if average_smurfing_amount >= 9000 and average_smurfing_amount < 10000:
    smurfing_risk_score += 40

# Large combined amount
if total_smurfing_amount >= 100000:
    smurfing_risk_score += 20

print("Smurfing Risk Score:", smurfing_risk_score)
```

```python
smurfing_alert = {
    "receiver_account": "TARGET01",
    "pattern": "SMURFING",
    "unique_senders": id_counts["TARGET01"],
    "total_amount": total_smurfing_amount,
    "average_transaction": average_smurfing_amount,
    "risk_score": smurfing_risk_score,
    "status": "HIGH RISK"
}

print(smurfing_alert)
```

```python
circular_data = [
    {
        "transaction_id": "C001",
        "sender_account": "CIRC_A",
        "receiver_account": "CIRC_B",
        "amount": 15000,
        "timestamp": "2026-08-13 15:01:00",
        "sender_ip": "10.10.0.1"
    },
    {
        "transaction_id": "C002",
        "sender_account": "CIRC_B",
        "receiver_account": "CIRC_C",
        "amount": 14800,
        "timestamp": "2026-08-13 15:03:00",
        "sender_ip": "10.10.0.2"
    },
    {
        "transaction_id": "C003",
        "sender_account": "CIRC_C",
        "receiver_account": "CIRC_A",
        "amount": 14500,
        "timestamp": "2026-08-13 15:05:00",
        "sender_ip": "10.10.0.3"
    }
]

circular_df = pd.DataFrame(circular_data)

print(circular_df)
```

```python
# Create sender → receiver relationships
connections = set(
    zip(
        circular_df["sender_account"],
        circular_df["receiver_account"]
    )
)

# Check whether the transactions form A → B → C → A
circular_detected = (
    ("CIRC_A", "CIRC_B") in connections
    and
    ("CIRC_B", "CIRC_C") in connections
    and
    ("CIRC_C", "CIRC_A") in connections
)

print("Circular Pattern Detected:", circular_detected)
```

```python
circular_risk_score = 0

# Circular flow detected
if circular_detected:
    circular_risk_score += 60

# Multiple transactions in the cycle
if len(circular_df) >= 3:
    circular_risk_score += 20

# Similar transaction amounts
amount_difference = circular_df["amount"].max() - circular_df["amount"].min()

if amount_difference <= 1000:
    circular_risk_score += 20

print("Circular Risk Score:", circular_risk_score)
```

```python
circular_alert = {
    "receiver_account": "CIRCULAR_NETWORK",
    "pattern": "CIRCULAR",
    "unique_senders": 3,
    "total_amount": circular_df["amount"].sum(),
    "average_transaction": circular_df["amount"].mean(),
    "risk_score": circular_risk_score,
    "status": "HIGH RISK"
}

print(circular_alert)
```

```python
import pandas as pd
import networkx as nx
import matplotlib.pyplot as plt

# Load transaction data
df = pd.read_excel("FinGraph_transactions (1).xlsx")

# Get Starburst transactions
starburst_df = df[df["receiver_account"] == "SHELL01"]

# Get the 50 sender accounts
suspicious_accounts = starburst_df["sender_account"].unique()

# Create directed graph
G = nx.DiGraph()

# Add sender → receiver connections
for account in suspicious_accounts:
    G.add_edge(account, "SHELL01")

print("Nodes:", G.number_of_nodes())
print("Edges:", G.number_of_edges())
```

```python
plt.figure(figsize=(12, 10))

pos = nx.spring_layout(G, seed=42)

nx.draw(
    G,
    pos,
    with_labels=True,
    node_size=800,
    arrows=True
)

plt.title("FinGraph - Starburst Fraud Pattern")
plt.show()
```

```python
plt.figure(figsize=(14, 10))

pos = nx.spring_layout(G, seed=42)

# Draw all nodes and edges
nx.draw_networkx_edges(G, pos, arrows=True)
nx.draw_networkx_nodes(
    G,
    pos,
    node_size=500
)

# Highlight suspicious receiver
nx.draw_networkx_nodes(
    G,
    pos,
    nodelist=["SHELL01"],
    node_size=1800
)

nx.draw_networkx_labels(G, pos, font_size=8)

plt.title("FinGraph - Starburst Fraud Network")
plt.axis("off")
plt.show()
```

```python
import networkx as nx
import matplotlib.pyplot as plt

G_all = nx.DiGraph()

# 1. STARBURST
for i in range(501, 551):
    G_all.add_edge(f"A{i}", "SHELL01")

# 2. SMURFING
for i in range(1, 21):
    G_all.add_edge(f"SMURF{i:03d}", "TARGET01")

# 3. FAN-OUT
for i in range(1, 16):
    G_all.add_edge("FANOUT01", f"ACC{i:03d}")

# 4. CIRCULAR
G_all.add_edge("CIRC_A", "CIRC_B")
G_all.add_edge("CIRC_B", "CIRC_C")
G_all.add_edge("CIRC_C", "CIRC_A")

print("Total Nodes:", G_all.number_of_nodes())
print("Total Edges:", G_all.number_of_edges())
```

```python
plt.figure(figsize=(16, 12))

pos = nx.spring_layout(G_all, seed=42, k=0.5)

nx.draw_networkx_edges(
    G_all,
    pos,
    arrows=True,
    alpha=0.5
)

nx.draw_networkx_nodes(
    G_all,
    pos,
    node_size=500
)

nx.draw_networkx_labels(
    G_all,
    pos,
    font_size=7
)

plt.title("FinGraph - Combined Fraud Network")
plt.axis("off")
plt.show()
```

```python

in_degree = dict(G_all.in_degree())
out_degree = dict(G_all.out_degree())

print("Top 10 nodes by incoming connections:")

sorted(in_degree.items(), key=lambda x: x[1], reverse=True)[:10]

```

```python
print("Top 10 nodes by outgoing connections:")

sorted(out_degree.items(), key=lambda x: x[1], reverse=True)[:10]
```

```python
network_risk = {}

for node in G_all.nodes():

    incoming = in_degree.get(node, 0)
    outgoing = out_degree.get(node, 0)

    score = 0

    # Many incoming connections → suspicious concentration
    if incoming >= 20:
        score += 50
    elif incoming >= 10:
        score += 30

    # Many outgoing connections → fan-out behavior
    if outgoing >= 10:
        score += 50
    elif outgoing >= 5:
        score += 30

    network_risk[node] = min(score, 100)

sorted(network_risk.items(), key=lambda x: x[1], reverse=True)[:10]
```

```python
network_risk_df = pd.DataFrame(
    list(network_risk.items()),
    columns=["account", "network_risk_score"]
)

network_risk_df = network_risk_df.sort_values(
    "network_risk_score",
    ascending=False
)

network_risk_df.head(10)
```

```python
import pandas as pd

alerts_df = pd.read_excel("FinGraph_Fraud_Alerts.xlsx")

print(alerts_df)
```

```python
final_alerts = alerts_df.copy()

final_alerts["network_risk_score"] = final_alerts["receiver_account"].map(
    network_risk
).fillna(0)

# Correct mapping for Fan-out
final_alerts.loc[
    final_alerts["pattern"] == "FAN_OUT",
    "network_risk_score"
] = network_risk.get("FANOUT01", 0)

# Correct mapping for Circular
final_alerts.loc[
    final_alerts["pattern"] == "CIRCULAR",
    "network_risk_score"
] = max(
    network_risk.get("CIRC_A", 0),
    network_risk.get("CIRC_B", 0),
    network_risk.get("CIRC_C", 0)
)

print(final_alerts)
```

```python
final_alerts["final_risk_score"] = (
    final_alerts["risk_score"] * 0.6
    + final_alerts["network_risk_score"] * 0.4
)

final_alerts["final_risk_score"] = final_alerts["final_risk_score"].round(2)

print(final_alerts)
```

```python
final_alerts.to_excel(
    "FinGraph_Final_Risk_Analysis.xlsx",
    index=False
)
```
